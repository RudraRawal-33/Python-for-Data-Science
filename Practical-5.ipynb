**B. Write a python code to show statistical information on given data set.**

```python
import statistics as st
```

```python
print(st.mean([1, 3, 5, 7, 9, 11, 13]))
print(st.median([-11, 5.5, -3.4, 7.1, -9, 22]))
print(st.mode([1, 1, 3, -5, 7, -9, 11]))
print(st.variance([1, 3, 5, 7, 9, 11]))
print(st.stdev([1, 30, 50, 100]))
```

**C.Write a python code to create pandas series from dictionaries.**

```python
dictionary = {'Niraj': 50, 'Harshal': 70, 'Rudra': 80}

series = pd.Series(dictionary, index=['Harshal', 'Rudra', 'Niraj'])

print("index parameter")
print(series)
```
