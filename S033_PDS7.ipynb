**7A. Write a Python program to create a scatter plot of Age and Height of 5 students.**

**7C. Write a Python program to create a histogram showing the distribution of marks of 10 students.**

```python
x = np.random.normal(0, 10, 100)

plt.hist(x)
plt.show()
```

**7D. Write a Python program to create a pie chart showing the number of students in different courses.**

```python
x = np.array([35, 25, 25, 15])
mylabels = ["Python", "OS", "DS", "AM"]

plt.pie(x, labels = mylabels)
plt.show()
```
